```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# -------------------------------
# 1. Create training dataset
# -------------------------------
train_data = {
    'Age': [18, 19, 20, 22, 28, 26, 24, 27],
    'Overall': [80, 85, 90, 78, 88, 86, 83, 87],
    'Potential': [85, 87, 92, 80, 90, 88, 85, 89],
    'Club': ['Chelsea', 'Manchester United', 'Manchester City', 'Arsenal', 'Tottenham Hotspurs', 'Chelsea', 'Manchester United', 'Manchester City'],
    'Position': ['ST', 'RW', 'CM', 'LW', 'RW', 'AM', 'RB', 'CAM'],
    'Value': [50e6, 70e6, 90e6, 45e6, 80e6, 75e6, 55e6, 78e6]  # Euros
}
df_train = pd.DataFrame(train_data)

# Encode categorical features
df_train_encoded = pd.get_dummies(df_train, columns=['Club', 'Position'], drop_first=True)
X_train = df_train_encoded.drop(columns=['Value'])
y_train = df_train_encoded['Value']

# -------------------------------
# 2. Train Decision Tree model
# -------------------------------
model = DecisionTreeRegressor(random_state=42)
model.fit(X_train, y_train)

# -------------------------------
# 3. Load your new player CSV
# -------------------------------
df_marketvalue = pd.read_csv('/Users/evansamoah/Desktop/Projects/Python/Prediction/newplayer_clean.csv')

# Remove spaces from column names just in case
df_marketvalue.columns = df_marketvalue.columns.str.strip()

# Encode categorical columns
df_marketvalue_encoded = pd.get_dummies(df_marketvalue, columns=['Club', 'Position'], drop_first=True)

# Add any missing columns from training
for col in X_train.columns:
    if col not in df_marketvalue_encoded.columns:
        df_marketvalue_encoded[col] = 0

# Match column order
df_marketvalue_encoded = df_marketvalue_encoded[X_train.columns]

# -------------------------------
# 4. Predict market values
# -------------------------------
predictions = model.predict(df_marketvalue_encoded)
df_marketvalue['PredictedValue (€)'] = predictions.round(2)

# -------------------------------
# 5. Display results
# -------------------------------
print(df_marketvalue)
```
